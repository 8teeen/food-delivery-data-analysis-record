```python
import pandas as pd

def convert_time(value):
    if pd.isna(value):
        return pd.NaT

    value = str(value).strip()

    # Handle Excel decimal time
    try:
        number = float(value)

        if 0 <= number <= 1:
            seconds = round(number * 24 * 60 * 60)
            return pd.to_timedelta(seconds, unit="s")

    except ValueError:
        pass

    # Handle normal time formats
    return pd.to_timedelta(value, errors="coerce")
```

```python
def convert_pickup_time(value):
    if pd.isna(value):
        return pd.NaT

    value = str(value).strip()

    # Excel decimal time
    try:
        number = float(value)
        if 0 <= number <= 1:
            total_seconds = round(number * 86400)
            return pd.to_timedelta(total_seconds, unit="s")
    except ValueError:
        pass

    # Convert clock time, including times beyond midnight
    parts = value.split(":")
    try:
        hours = int(parts[0])
        minutes = int(parts[1])
        seconds = int(parts[2]) if len(parts) == 3 else 0

        if minutes < 60 and seconds < 60:
            return pd.Timedelta(
                hours=hours,
                minutes=minutes,
                seconds=seconds
            )
    except (ValueError, IndexError):
        pass

    return pd.NaT
```
