We calculate descriptive statistics to understand
the overall delivery performance.

The Time_taken (min) column represents the total
delivery time in minutes.

```python
duration = df["Time_taken (min)"]
baseline = pd.Series({
    "Total Deliveries": duration.count(),
    "Mean (minutes)": duration.mean(),
    "Median (minutes)": duration.median(),
    "Minimum (minutes)": duration.min(),
    "Maximum (minutes)": duration.max(),
    "25th Percentile": duration.quantile(0.25),
    "75th Percentile": duration.quantile(0.75),
    "Standard Deviation": duration.std()
})

display(baseline.round(2).to_frame(name="Value"))
```

```python
print("Skewness:", round(duration.skew(), 2))

Q1 = duration.quantile(0.25)
Q3 = duration.quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("IQR:", IQR)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)


outliers = duration[
    (duration < lower_limit) | (duration > upper_limit)
]

print("Potential outliers:", len(outliers))
```

## 4. Traffic Analysis

In this section, we analyze how road traffic density
is associated with food delivery duration.

We compare the number of deliveries, average delivery
time, and median delivery time across traffic categories.

```python
print("Traffic Categories:")
print(df["Road_traffic_density"].unique())

print("\nDelivery Counts:")
print(df["Road_traffic_density"].value_counts(dropna=False))
```

```python
traffic_analysis = df.groupby(
    "Road_traffic_density"
)["Time_taken (min)"].agg(
    Delivery_Count="count",
    Mean_Duration="mean",
    Median_Duration="median"
)
traffic_analysis = traffic_analysis.round(2)

display(traffic_analysis)
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=df,
    x="Road_traffic_density",
    y="Time_taken (min)",
    errorbar=None
)

plt.title("Average Delivery Duration by Traffic Density")
plt.xlabel("Road Traffic Density")
plt.ylabel("Average Delivery Duration (minutes)")

plt.show()
```

```python

**Question:** Does road traffic density relate to delivery duration?

**Result:**
The Jam traffic category recorded the highest mean
delivery duration of 31.18 minutes across 14,139 deliveries.

The Low traffic category recorded the lowest mean
delivery duration of 21.27 minutes across 15,476 deliveries.

The difference between these two categories was
9.91 minutes.

**Interpretation:**
The results show that deliveries in the Jam traffic
category were associated with longer delivery times
compared with deliveries in the Low traffic category.

Traffic conditions may be one factor associated with
delivery duration. However, other factors such as distance,
weather, and delivery-person experience may also influence
the observed differences.

**Data limitation:**
There were 601 records with missing traffic information.
These records were excluded from the traffic comparison.
```

We use the Haversine formula to calculate the
approximate distance between the restaurant and
delivery location using their latitude and longitude.

The calculated distance will be stored in a new
column called Distance_km.

```python
import numpy as np
lat1 = np.radians(df["Restaurant_latitude"])
lon1 = np.radians(df["Restaurant_longitude"])

lat2 = np.radians(df["Delivery_location_latitude"])
lon2 = np.radians(df["Delivery_location_longitude"])

dlat = lat2 - lat1
dlon = lon2 - lon1

a = (
    np.sin(dlat / 2)**2
    + np.cos(lat1) * np.cos(lat2)
    * np.sin(dlon / 2)**2
)

a = np.clip(a, 0, 1)

c = 2 * np.arcsin(np.sqrt(a))

earth_radius = 6371

df["Distance_km"] = earth_radius * c

print("Distance calculated successfully!")
```

```python
# Import NumPy
import numpy as np

# Convert coordinates from degrees to radians
lat1 = np.radians(df["Restaurant_latitude"])
lon1 = np.radians(df["Restaurant_longitude"])

lat2 = np.radians(df["Delivery_location_latitude"])
lon2 = np.radians(df["Delivery_location_longitude"])

# Calculate differences between coordinates
dlat = lat2 - lat1
dlon = lon2 - lon1

# Apply Haversine formula
a = (
    np.sin(dlat / 2)**2
    + np.cos(lat1) * np.cos(lat2)
    * np.sin(dlon / 2)**2
)

# Protect against small floating-point errors
a = np.clip(a, 0, 1)

c = 2 * np.arcsin(np.sqrt(a))

# Earth's approximate radius in kilometers
earth_radius = 6371

# Calculate distance
df["Distance_km"] = earth_radius * c

print("Distance calculated successfully!")
```


### Key Observations

1. Road traffic density: Deliveries during Jam traffic took an average of 31.18 minutes, compared with 21.27 minutes during Low traffic.

2. Weather conditions: Cloudy weather had the highest average delivery time (28.92 minutes), while Sunny weather had the lowest (21.86 minutes).

3. Vehicle type: Motorcycles had the highest average delivery time (27.61 minutes), while Electric Scooters had the lowest (24.47 minutes).

4. Multiple deliveries: Average delivery time increased from 22.88 minutes with zero additional deliveries to 47.82 minutes with three additional deliveries.

5. Delivery ratings: Ratings between 3.5 and 3.9 generally corresponded to longer average delivery times than ratings between 4.5 and 4.9.

### Conclusion

Exploratory Data Analysis revealed relationships between delivery duration, traffic density, weather, vehicle type, multiple deliveries, and delivery person ratings.

These findings can help identify factors associated with longer delivery times and guide further analysis.
