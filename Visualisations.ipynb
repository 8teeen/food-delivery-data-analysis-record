
## 4. Delivery Distance Calculation
The dataset contains restaurant and delivery location coordinates in latitude and longitude format.
To calculate the distance between these two locations, we will use the Haversine formula. This formula calculates the approximate geographical distance between two points on Earth.
The calculated distance will be measured in kilometers and used in the Distance vs Delivery Duration scatter plot.


```python
coordinate_columns = [
    'Restaurant_latitude',
    'Restaurant_longitude',
    'Delivery_location_latitude',
    'Delivery_location_longitude' ]
print("Missing Coordinate Values:")
print(df[coordinate_columns].isnull().sum())
print("\nCoordinate Data Types:")
print(df[coordinate_columns].dtypes)
```

```python
lat1 = np.radians(df['Restaurant_latitude'])
lon1 = np.radians(df['Restaurant_longitude'])
lat2 = np.radians(df['Delivery_location_latitude'])
lon2 = np.radians(df['Delivery_location_longitude'])

dlat = lat2 - lat1
dlon = lon2 - lon1

a = (np.sin(dlat / 2) ** 2
    + np.cos(lat1) * np.cos(lat2)
    * np.sin(dlon / 2) ** 2 )
c = 2 * np.arcsin(np.sqrt(a))
earth_radius = 6371
df['Distance_km'] = earth_radius * c
print("Delivery distance calculated successfully!")
```



A histogram is used to understand how delivery durations are distributed across all deliveries.
This visualization helps identify the most common delivery-time range, the overall distribution shape, and any unusually short or long delivery durations.


```python

duration = df['Time_taken (min)'].dropna()

print("Delivery Duration Statistics")
print("----------------------------")
print("Total Deliveries:", len(duration))
print("Mean:", round(duration.mean(), 2), "minutes")
print("Median:", round(duration.median(), 2), "minutes")
print("Minimum:", duration.min(), "minutes")
print("Maximum:", duration.max(), "minutes")
print("Standard Deviation:", round(duration.std(), 2), "minutes")
```

```python

plt.figure(figsize=(10, 6))
sns.histplot(
    duration,
    bins=30,
    kde=True,
    color='steelblue')

plt.title(
    'Distribution of Food Delivery Duration',
    fontsize=16,
    fontweight='bold')

plt.xlabel('Delivery Duration (minutes)', fontsize=12)
plt.ylabel('Number of Deliveries', fontsize=12)

plt.tight_layout()
plt.savefig(
    'visualizations/01_delivery_duration_distribution.png',
    dpi=300,
    bbox_inches='tight')
plt.show()
```



The histogram shows the distribution of delivery durations across the dataset. Most deliveries are concentrated around [actual time range] minutes, indicating the most common delivery duration range. The distribution shows [observed shape], although extreme values and the use of fixed histogram bins may affect how the pattern is interpreted.


```python

# Create 2-minute histogram intervals

plt.figure(figsize=(12, 6))

# Create bins with exactly 2-minute widths
bin_edges = np.arange(
    np.floor(duration.min() / 2) * 2,
    np.ceil(duration.max() / 2) * 2 + 2,
    2
)

sns.histplot(
    duration,
    bins=bin_edges,
    kde=False,
    color='red',
    edgecolor='black'
)

plt.title(
    'Distribution of Food Delivery Duration',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Delivery Duration (minutes)', fontsize=12)
plt.ylabel('Number of Deliveries', fontsize=12)

plt.xticks(bin_edges, rotation=45)

plt.tight_layout()

plt.savefig(
    'visualizations/01_delivery_duration_distribution.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```


## 6. Traffic Comparison – Median Delivery Duration

This visualization compares the median delivery duration across different road traffic categories.
A bar chart is used to identify differences in delivery times under Low, Medium, High, and Jam traffic conditions.
The median is used instead of the mean because it is less affected by unusually high or low delivery durations.


```python


traffic_analysis = df.groupby('Road_traffic_density').agg(
    Median_Duration=('Time_taken (min)', 'median'),
    Total_Deliveries=('Time_taken (min)', 'count')
).reset_index()
print("Traffic Category Analysis:")
print(traffic_analysis)
```

```python

# Calculate median delivery duration and group sizes by traffic category

traffic_analysis = df.groupby('Road_traffic_density').agg(
    Median_Duration=('Time_taken (min)', 'median'),
    Total_Deliveries=('Time_taken (min)', 'count')
)

print("Traffic Category Analysis:")
print(traffic_analysis)
```

```python


plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=traffic_analysis,
    x='Road_traffic_density',
    y='Median_Duration',
    color='light_green')

for container in ax.containers:
    ax.bar_label(
        container,
        fmt='%.2f',
        padding=5 )
plt.title(
    'Median Delivery Duration by Traffic Category',
    fontsize=16,
    fontweight='bold')

plt.xlabel('Road Traffic Density', fontsize=12)
plt.ylabel('Median Delivery Duration (minutes)', fontsize=12)
plt.tight_layout()
plt.savefig(
    'visualizations/02_traffic_comparison.png',
    dpi=300,
    bbox_inches='tight')
plt.show()
```


## 7. Weather Comparison – Average Delivery Duration

This visualization compares the average delivery duration across different weather conditions.
A bar chart is used to identify which weather categories have higher or lower observed average delivery durations. The comparison helps us understand how delivery times differ across weather conditions.


```python


weather_analysis = df.groupby('Weather_conditions').agg(
    Average_Duration=('Time_taken (min)', 'mean'),
    Median_Duration=('Time_taken (min)', 'median'),
    Total_Deliveries=('Time_taken (min)', 'count')
).reset_index()

print("Weather Category Analysis:")
print(weather_analysis.round(2))
```

```python


plt.figure(figsize=(12, 6))
ax = sns.barplot(
    data=weather_analysis,
    x='Weather_conditions',
    y='Average_Duration',
    color='steelblue')
for container in ax.containers:
    ax.bar_label(
        container,
        fmt='%.2f',
        padding=5)

plt.title(
    'Average Delivery Duration by Weather Condition',
    fontsize=16,
    fontweight='bold')

plt.xlabel('Weather Conditions', fontsize=12)
plt.ylabel('Average Delivery Duration (minutes)', fontsize=12)
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig(
    'visualizations/03_weather_comparison.png',
    dpi=300,
    bbox_inches='tight')

plt.show()
```

```python
plt.figure(figsize=(12, 6))
ax = sns.barplot(
    data=weather_analysis,
    x='Weather_conditions',
    y='Average_Duration',
    color='steelblue')
for container in ax.containers:
    ax.bar_label(
        container,
        fmt='%.2f',
        padding=3)

plt.title(
    'Average Delivery Duration by Weather Condition',
    fontsize=16,
    fontweight='bold')

plt.xlabel('Weather Conditions', fontsize=12)
plt.ylabel('Average Delivery Duration (minutes)', fontsize=12)
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig(
    'visualizations/03_weather_comparison.png',
    dpi=300,
    bbox_inches='tight')

plt.show()
```

### Interpretation

The chart shows that Cloudy weather has the highest average delivery duration at 28.92 minutes, closely followed by Fog at 28.91 minutes. Sunny weather has the lowest average delivery duration at 21.86 minutes. These results show an association between weather conditions and delivery duration, but do not prove that weather alone causes delays. Also, 616 records with missing weather classifications are excluded from this comparison.
